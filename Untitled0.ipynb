#Visualizations

```python
import matplotlib.pyplot as plt
import seaborn as sns
import pandas as pd

df=pd.read_csv("zomato.csv" , encoding='latin1')
top_cities = df['City'].value_counts().head(10)

plt.figure(figsize=(12,6))

top_cities.plot(kind='bar')

plt.title("Top 10 Cities with Most Restaurants")
plt.xlabel("City")
plt.ylabel("Number of Restaurants")

plt.xticks(rotation=45)

plt.show()
```

```python
top_cusines = df['Cuisines'].value_counts().head(10)

plt.figure(figsize=(12,6))

top_cusines.plot(kind='bar')

plt.title("Top 10 Cuisines with Most Restaurants")
plt.xlabel("Cuisines")
plt.ylabel("Number of Restaurants")

plt.xticks(rotation=45)

plt.show()
```

```python
top_rating = df['Aggregate rating'].value_counts().head(10)

plt.figure(figsize=(12,6))

top_rating.plot(kind='bar')

plt.title("Number of Restaurants with most rating")
plt.xlabel("rating")
plt.ylabel("Number of Restaurants")

plt.xticks(rotation=45)

plt.show()
```

```python
delivery = df['Has Online delivery'].value_counts()

plt.figure(figsize=(6,6))

plt.pie(
    delivery,
    labels=delivery.index,
    autopct='%1.1f%%'
)

plt.title("Online Delivery Availability")

plt.show()
```

```python
plt.figure(figsize=(10,6))

plt.scatter(
    df['Votes'],
    df['Aggregate rating']
)

plt.title("Votes vs Aggregate Rating")

plt.xlabel("Votes")
plt.ylabel("Aggregate Rating")

plt.show()
```

```python
numeric_df = df.select_dtypes(include=['float64', 'int64'])

plt.figure(figsize=(10,6))

sns.heatmap(
    numeric_df.corr(),
    annot=True,
    cmap='coolwarm'
)

plt.title("Correlation Heatmap")

plt.show()
```

```python
plt.figure(figsize=(10,6))

plt.hist(
    df['Aggregate rating'],
    bins=10
)

plt.title("Distribution of Aggregate Ratings")

plt.xlabel("Ratings")
plt.ylabel("Number of Restaurants")

plt.show()
```

```python
price_rating = df.groupby('Price range')[
    'Aggregate rating'
].mean()

plt.figure(figsize=(8,5))

plt.plot(
    price_rating.index,
    price_rating.values,
    marker='o'
)

plt.title("Average Rating by Price Range")

plt.xlabel("Price Range")
plt.ylabel("Average Rating")

plt.grid(True)

plt.show()
```

1. New Delhi has the highest number of restaurants

The analysis shows that New Delhi contains the largest concentration of restaurants in the dataset, significantly higher than other cities. This suggests that Delhi has a highly competitive and diverse food market.

2. Most restaurants have ratings between 3.0 and 4.5

The histogram analysis indicates that the majority of restaurants fall within the average-to-good rating range. Very few restaurants have extremely low or perfect ratings.

3. Online delivery is not available in many restaurants

The pie chart shows that a large percentage of restaurants do not provide online delivery services. This indicates potential growth opportunities for delivery platforms.
